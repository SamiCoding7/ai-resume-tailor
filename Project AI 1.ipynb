```python
import google.generativeai as genai
import os
from dotenv import load_dotenv
load_dotenv("/Users/samiksha/ai-resume-tailor/.env")
api_key = os.getenv("GEMINI_API_KEY")
genai.configure(api_key=api_key)

for model in genai.list_models():
    print(model.name)

# Load the Gemini model
model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
import PyPDF2
chat = model.start_chat()

inputResume = input("Upload your resume as a .txt file")
inputDescription = input("Paste the job description")

def readFile(filepath):
    with open(filepath, "r") as file:
        text = file.read()
    return text
def readPdf(filepath):
    with open(filepath, "rb") as file:
        reader = PyPDF2.PdfReader(file)
        strTotal = ""
        for page in reader.pages:
            strTotal += page.extract_text()
    return strTotal

def processResume(resumeText,inputDescription):
    print("Resume Upload successful!")
    print("Description Upload successful!")
    response = chat.send_message(f"Give feedback on this resume: {resumeText} where this is the job description: {inputDescription}. Give it in four sections in this format: 1. A match score as a %, 2. Rewritten bullet points tailored to the job, 3. skill gaps what the job wants that the resume is missing, 4. key words to add-terms from the job description not in the resume")
    feedback = response.text
    print(feedback)
    
    #Save feedback to a file
    fileOutputPath = input("Please enter a file path to save feedback (use .txt extension e.g. /Users/yourname/Desktop/feedback.txt): ")
    with open(fileOutputPath, "w") as file:
        file.write(feedback)
    print(f"Feedback saved to {fileOutputPath}")
    
if inputResume.endswith(".txt"):
    try:
        resumeText = readFile(inputResume)
        processResume(resumeText,inputDescription)
    except FileNotFoundError:
        print("File not found, please check your path and try again.")
elif inputResume.endswith(".pdf"):
    try:
        resumeText = readPdf(inputResume)      
        processResume(resumeText,inputDescription)
    except FileNotFoundError:
        print("File not found, please check your path and try again.")
else:
    print("Please only upload a .txt or .pdf file")

    

```
